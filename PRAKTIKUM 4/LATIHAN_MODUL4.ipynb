```python
# PROGRAM MENGHITUNG TOTAL PEMBELIAN

n = int(input('MASUKKAN JUMLAH JENIS BARANG: '))

total = 0

for i in range(1, n+1):
    print('\nBARANG KE-', i)
    harga = int(input('MASUKKAN HARGA BARANG: '))
    jumlah = int(input('MASUKKAN JUMLAH BARANG: '))

    subtotal = harga * jumlah
    total = total + subtotal

    print('SUBTOTAL = ', subtotal)

print('\nTOTAL PEMBELIAN = ', total)
```

```python
# PROGRAM MENGHITUNG BILANGAN GENAP DAN GANJIL

n = int(input('MASUKKAN BANYAKNYA BILANGAN: '))

genap = 0
ganjil = 0

for i in range(1, n+1):
    bilangan = int(input('MASUKKAN BILANGAN: '))

    if bilangan % 2 == 0:
        genap = genap + 1
    else:
        ganjil = ganjil + 1

print('\nJUMLAH BILANGAN GENAP = ', genap)
print('JUMLAH BILANGAN GANJIL = ', ganjil)
```

```python
# PROGRAM BILANGAN KELIPATAN 3

awal = int(input('MASUKKAN NILAI AWAL: '))
akhir = int(input('MASUKKAN NILAI AKHIR: '))

print('BILANGAN KELIPATAN 3:')

for i in range(awal, akhir+1):
    if i % 3 == 0:
        print(i, end=' ')
```
