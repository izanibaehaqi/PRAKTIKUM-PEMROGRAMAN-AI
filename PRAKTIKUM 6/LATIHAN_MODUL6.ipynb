```python
#Contoh penggunaan Nested Loop

i = 5
while(i < 100):
    j = 3
    while(j <= (i/j)):
        if not(i%j): break
        j = j + 1
    if (j > i/j) : print(i, "MUSIKKK IS FUN")
    i = i + 1

print("UKM MUSIK POLTEK PURBAYA")
```
